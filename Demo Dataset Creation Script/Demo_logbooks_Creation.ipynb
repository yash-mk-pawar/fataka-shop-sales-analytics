```python

# Diwali dates for the past 3 logbook years (Years 1, 2, 3)
DIWALI_DATES = {
    1: datetime(2023, 11, 12),  # Year 1 (2023)
    2: datetime(2024, 11, 1),   # Year 2 (2024)
    3: datetime(2025, 10, 20)   # Year 3 (2025)
}
```

```python

def generate_large_logbook(year_num, diwali_date):
    records = []
    # Span across a 60-day festive season (45 days before to 15 days after Diwali)
    start_date = diwali_date - timedelta(days=45)
    end_date = diwali_date + timedelta(days=15)
    
    tx_id_counter = 100000 * year_num
    current_date = start_date
    
    while current_date <= end_date:
        days_to_diwali = (diwali_date - current_date).days
        
        # Adjust daily transaction volume based on proximity to Diwali
        if -2 <= days_to_diwali <= 3:
            daily_tx_count = random.randint(160, 220) # Peak festival rush
        elif -15 <= days_to_diwali < -2:
            daily_tx_count = random.randint(80, 130)  # Pre-Diwali buildup
        else:
            daily_tx_count = random.randint(25, 55)   # Early/late season trickle
            
        for _ in range(daily_tx_count):
            tx_id_counter += 1
            prod = random.choice(PRODUCTS)
            quantity_sold = random.randint(1, 6)
            total_amount = quantity_sold * prod["price"]
            
            records.append({
                "Transaction_ID": f"TX{tx_id_counter}",
                "Date": current_date.strftime("%Y-%m-%d"),
                "Days_To_Diwali": days_to_diwali,
                "Product_ID": prod["product_id"],
                "Product_Name": prod["name"],
                "Category": prod["category"],
                "Unit_Price": prod["price"],
                "Quantity_Sold": quantity_sold,
                "Total_Revenue": total_amount,
                "Data_Source": f"Logbook_Year_{year_num}"
            })
            
        current_date += timedelta(days=1)
        
    df = pd.DataFrame(records)
    
    # Ensure exact record count constraint (between 3,000 and 4,000 records)
    if len(df) > 4000:
        df = df.sample(n=4000, random_state=42).sort_values("Date").reset_index(drop=True)
    elif len(df) < 3000:
        extra = df.sample(n=3000 - len(df), replace=True, random_state=42)
        df = pd.concat([df, extra], ignore_index=True).sort_values("Date").reset_index(drop=True)
        
    filename = f"../Dataset/diwali_sales_year_{year_num}.csv"
    df.to_csv(filename, index=False)
    print(f"Generated {filename} with {len(df):,} transaction records.")
```

```python
if __name__ == "__main__":
    for yr, d_date in DIWALI_DATES.items():
        generate_large_logbook(yr, d_date)
    print("All 3 large-scale logbook CSV files successfully created!")
```
